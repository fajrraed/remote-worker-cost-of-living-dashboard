```python
import pandas as pd

# to Load the dataset
df = pd.read_csv("global_city_col_qol_index.csv")

#  to Print basic information about rows, columns, and data types
print("--- Dataset Summary ---")
df.info()

# to View the first 5 rows to see how the actual values look
print("\n--- First 5 Rows ---")
df.head()
```

```python
# to Check how many missing values exist in each column
missing_data = df.isnull().sum()
print("--- Missing Values Per Column ---")
print(missing_data[missing_data > 0]) # Shows only columns that have missing data
```

```python
# to Sort by city and survey_date so the most recent dates are at the bottom
df_sorted = df.sort_values(by=['city', 'survey_date'])

# to Drop duplicates for the same city, keeping only the LAST (most recent) record
df_latest = df_sorted.drop_duplicates(subset=['city'], keep='last').copy()

# to Drop rows where internet_speed_mbps is missing, as remote workers require this data
df_clean = df_latest.dropna(subset=['internet_speed_mbps']).copy()

# to Drop the 'analyst_note' column since it's mostly empty (193 missing values)
if 'analyst_note' in df_clean.columns:
    df_clean = df_clean.drop(columns=['analyst_note'])

# to Check the new size of our clean dataset
print(f"Original rows: {len(df)}")
print(f"Cleaned unique cities remaining: {len(df_clean)}")
```

```python
# to Export the clean dataframe to a new CSV file without the index column
df_clean.to_csv("cleaned_nomad_cities.csv", index=False)
print("Success! 'cleaned_nomad_cities.csv' has been saved and is ready for Power BI.")
```
